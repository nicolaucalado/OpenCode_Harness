## 1. Variáveis e tipos

Python é dinamicamente tipado, mas tipos importam: `int`, `float`, `str`, `bool` e `None` cobrem a maior parte do dia a dia. Use `type()` para inspecionar e f-strings para formatar.


```python
nome = "Nicolau"
idade = 30
altura = 1.78
estudando = True
apelido = None

for valor in (nome, idade, altura, estudando, apelido):
    print(f"{valor!r:>10} -> {type(valor).__name__}")

print(f"{nome} tem {idade} anos e {altura:.2f} m de altura")
```

## 2. Operadores

```python
a, b = 17, 5

print("soma        ", a + b)
print("divisão     ", a / b)
print("divisão int ", a // b)
print("resto       ", a % b)
print("potência    ", a**2)

print("maior de idade?", idade >= 18)
print("entre 18 e 65?", 18 <= idade < 65)
print("precisa estudar?", not estudando)

texto = "  Análise de Dados com Python  "
print(repr(texto.strip()))
print(texto.strip().lower().replace("python", "pandas"))
print("palavras:", len(texto.split()))
```

## 4. Controle de fluxo

`if/elif/else` para decisões, `for` para percorrer, `while` quando a parada depende de uma condição. `enumerate` dá índice e valor; `zip` percorre duas listas juntas; *comprehension* cria listas em uma linha.


```python
for nota in (5.9, 6.0, 7.5, 9.1):
    if nota >= 9:
        conceito = "A"
    elif nota >= 7:
        conceito = "B"
    elif nota >= 6:
        conceito = "C"
    else:
        conceito = "D"
    print(f"{nota:.1f} -> {conceito}")

produtos = ["notebook", "mouse", "teclado"]
precos = [3500.0, 80.0, 150.0]
for i, (produto, preco) in enumerate(zip(produtos, precos, strict=True), start=1):
    print(f"{i}. {produto:<8} R$ {preco:>8.2f}")

tentativas = 0
while True:
    tentativas += 1
    if tentativas == 3:
        break
print("tentativas:", tentativas)

caros = [preco for preco in precos if preco > 100]
print("caros:", caros)
```

## 5. Funções

Função pequena, com nome claro, type hints e docstring. Valor padrão evita repetir argumentos; `ValueError` comunica entrada inválida melhor do que retornar `None` silenciosamente.


```python
def media(valores: list[float]) -> float:
    """Média aritmética simples; protege contra lista vazia."""
    if not valores:
        raise ValueError("lista vazia não tem média")
    return sum(valores) / len(valores)


def resumo(valores: list[float], casas: int = 2) -> dict[str, float]:
    """Retorna mínimo, máximo e média arredondados."""
    return {
        "min": round(min(valores), casas),
        "max": round(max(valores), casas),
        "media": round(media(valores), casas),
    }


print(resumo(notas))
print(sorted(perfil.items(), key=lambda item: item[0]))
```

## 6. Mini-análise sem pandas

Antes de importar biblioteca, resolva com Python puro: dá para ver de onde vem cada número que o pandas vai calcular depois. O padrão `dict.get(chave, 0) + valor` é o mesmo de um `groupby().sum()`.


```python
vendas = [
    {"mes": "2026-08", "produto": "notebook", "valor": 3500.0, "qtd": 2},
    {"mes": "2026-08", "produto": "mouse", "valor": 80.0, "qtd": 10},
    {"mes": "2026-09", "produto": "notebook", "valor": 3500.0, "qtd": 1},
    {"mes": "2026-09", "produto": "teclado", "valor": 150.0, "qtd": 4},
    {"mes": "2026-10", "produto": "mouse", "valor": 80.0, "qtd": 6},
]

total = sum(item["valor"] * item["qtd"] for item in vendas)
unidades = sum(item["qtd"] for item in vendas)

por_produto: dict[str, float] = {}
for item in vendas:
    por_produto[item["produto"]] = (
        por_produto.get(item["produto"], 0.0) + item["valor"] * item["qtd"]
    )

top = max(por_produto.items(), key=lambda par: par[1])

print(f"Faturamento total: R$ {total:.2f}")
print(f"Unidades vendidas: {unidades}")
print(f"Ticket médio:      R$ {total / unidades:.2f}")
print(f"Produto líder:     {top[0]} (R$ {top[1]:.2f})")
```

## 7. Exercícios

Implemente as duas funções abaixo seguindo a docstring. Não altere os testes da célula seguinte — eles são o critério de aceite.

1. `classificar_ticket(valor)` → `"baixo"` se `< 100`, `"medio"` se `< 500`, senão `"alto"`. Valor negativo deve levantar `ValueError`.
2. `aplicar_desconto(valor, percentual)` → valor final arredondado em 2 casas. Percentual fora de `0..100` deve levantar `ValueError`.


```python
def classificar_ticket(valor: float) -> str:
    """Classifica um ticket em 'baixo' (<100), 'medio' (<500) ou 'alto'."""
    if valor < 0:
        raise ValueError("valor não pode ser negativo")
    if valor < 100:
        return "baixo"
    if valor < 500:
        return "medio"
    return "alto"


def aplicar_desconto(valor: float, percentual: float) -> float:
    """Aplica desconto percentual (0 a 100) e devolve o valor final."""
    if not 0 <= percentual <= 100:
        raise ValueError("percentual deve estar entre 0 e 100")
    return round(valor * (1 - percentual / 100), 2)
```

```python
assert classificar_ticket(99.99) == "baixo"
assert classificar_ticket(100) == "medio"
assert classificar_ticket(500) == "alto"

assert aplicar_desconto(200, 10) == 180.0
assert aplicar_desconto(100, 0) == 100.0
assert aplicar_desconto(100, 100) == 0.0

for chamada in (lambda: classificar_ticket(-1), lambda: aplicar_desconto(100, 101)):
    try:
        chamada()
    except ValueError:
        pass
    else:
        raise AssertionError("entrada inválida deveria levantar ValueError")

print("exercícios OK")
```
